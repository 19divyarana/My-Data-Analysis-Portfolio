```python
import numpy as np

sales = np.array([
    [101, 1200, 200, 5],
    [102, 1500, 300, 8],
    [103, 1800, 400, 10],
    [104, 1000, 150, 4],
    [105, 2200, 500, 12]
])
```

```python
total_revenue = np.sum(sales[:,1]) #1. Find total revenue
print(total_revenue)
```

```python
total_profit = np.sum(sales[:,2]) #2. Find total profit
print(total_profit)
```

```python
avg_revenue = np.mean(sales[:,1])#3. Find average revenue
print(avg_revenue)
```

```python
avg_profit = np.mean(sales[:,2]) #4. Find average profit
print(avg_profit)
```

```python
highest_revenue = np.max(sales[:,1]) #5. Find highest revenue
print(highest_revenue)
```

```python
lowest_revenue = np.min(sales[:,1])#6. Find lowest revenue
print(lowest_revenue)
```

```python
index = np.argmax(sales[:,2]) #7. Find customer with maximum profit
customer = sales[index,0]

print(customer)
```

```python
index = np.argmin(sales[:,2]) #8. Find customer with minimum profit
customer = sales[index,0]

print(customer)
```

```python
profit_margin = (sales[:,2] / sales[:,1]) * 100 #9. Calculate profit margin for each customer
print(profit_margin)
```

```python
total_orders = np.sum(sales[:,3]) #10. Find total orders
print(total_orders)
```

```python
avg_orders = np.mean(sales[:,3]) #11. Find average orders
print(avg_orders)
```

```python
revenue_per_order = sales[:,1] / sales[:,3] #12. Find revenue generated per order
print(revenue_per_order)
```

```python
high_revenue_customers = sales[sales[:,1] > 1500] #13. Find customers with revenue greater than 1500

print(high_revenue_customers)
```

```python
count = np.sum(sales[:,2] > 300) #14. Count customers with profit above 300

print(count)
```

```python
std_revenue = np.std(sales[:,1]) #15. Find standard deviation of revenue

print(std_revenue)
```

```python
sorted_sales = sales[sales[:,1].argsort()] #16. Sort customers by revenue

print(sorted_sales)
```

```python
median_revenue = np.median(sales[:,1]) #17. Find median revenue

print(median_revenue)
```

```python
cumulative_revenue = np.cumsum(sales[:,1]) #18. Calculate cumulative revenue

print(cumulative_revenue)
```

```python
correlation = np.corrcoef(sales[:,1], sales[:,2]) #19. Find correlation between revenue and profit

print(correlation)
```
