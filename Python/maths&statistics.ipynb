```python
import numpy as np
arr = np.arange(1, 11)

print(np.sqrt(arr))
print(np.exp(arr))
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.mean(arr))    # 30.0
print(np.median(arr))  # 30.0
print(np.std(arr))     # 14.14 (approx)
```

```python
arr = np.array([[1, 5, 3], [4, 2, 6]])

print(np.min(arr, axis=1))  # [1 2]
print(np.max(arr, axis=1))  # [5 6]
```

```python
arr = np.array([1, 10, 100])

print(np.log(arr))
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print(np.sum(arr, axis=0))  # [5 7 9]
```

```python
arr = np.random.rand(100)

print(np.mean(arr))  # ~0.5 (random)
print(np.var(arr))   # ~0.08 (random)
```

```python
arr = np.array([10, 20, 30, 40, 50])

z = (arr - np.mean(arr)) / np.std(arr)
print(z)
```
